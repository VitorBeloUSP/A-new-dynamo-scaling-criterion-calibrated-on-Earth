```python
df=pd.read_csv('sigma_todos_exoplanetas.csv')
```

```python
def campo_radial(sigma, D, rho, R_m, r, K):
    
    Br=((K*R_m**2*D**2)/(2*sigma*r**3))*(np.sqrt(rho/mi_0)) #Tesla

    return Br
```

```python
Rd=0.55*R_t # m
Rdi=0.19*R_t # m
rho=11000 # kg/m^3
D=Rd-Rdi # m
mi_0=4*np.pi*10**(-7)
R_m_terra=800 #ad
B_terra=5*10**(-5) # Tesla
sigma_terra=1.5*10**6 #(S/m)
```

```python
K1=np.around(((2*B_terra*sigma_terra*R_t**3)/(R_m_terra**2*D**2))*(np.sqrt(mi_0/rho)), 3)
```
