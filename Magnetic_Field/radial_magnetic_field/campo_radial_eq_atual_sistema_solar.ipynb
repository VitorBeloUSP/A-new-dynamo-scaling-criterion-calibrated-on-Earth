```python
def campo_radial(sigma, D, rho, R_m, r, K):
    
    Br=((K*R_m**2*D**2)/(2*sigma*r**3))*(np.sqrt(rho/mi_0)) #Tesla

    return Br
```

```python
def fmt(x):
    if isinstance(x, list):
        return [f'{v:.2e}' for v in x]
    elif isinstance(x, (float, np.floating)) and not np.isnan(x) and x != 0.0:
        return f'{x:.2e}'
    else:
        return x

df = pd.DataFrame(resultados)  # recria do dicionário original
df_fmt = df.map(fmt)
print(df_fmt.to_string(index=False))
```
